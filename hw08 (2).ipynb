```python
# HW08：Python 聚餐費用試算器
# 姓名：廖音淇
# 學號：D1370904

# 共同的原始資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"
service_rate = 0.10
budget_per_person = 300

# 共同的前置處理
cleaned_name = event_name.strip().lower()
meal_cost = float(meal_amount.replace("NT$", "").replace(",", "").strip())
service_fee = meal_cost * service_rate

print("=== 測試一：4人聚餐，折價100元 ===")
discount_1 = 100
num_people_1 = int(people)
total_amount_1 = meal_cost + service_fee - discount_1
per_person_cost_1 = total_amount_1 / num_people_1
is_over_budget_1 = per_person_cost_1 > budget_per_person
print(f"聚餐名稱：{cleaned_name}")
print(f"應付總額：{total_amount_1:.2f} 元")
print(f"每人費用：{per_person_cost_1:.2f} 元")
print(f"是否超過預算：{is_over_budget_1}\n")

print("=== 測試二：5人聚餐，折價100元 ===")
num_people_2 = 5
total_amount_2 = meal_cost + service_fee - discount_1
per_person_cost_2 = total_amount_2 / num_people_2
is_over_budget_2 = per_person_cost_2 > budget_per_person
print(f"分攤人數：{num_people_2} 人")
print(f"每人費用：{per_person_cost_2:.2f} 元")
print(f"是否超過預算：{is_over_budget_2}\n")

print("=== 測試三：4人聚餐，折價220元 ===")
discount_3 = 220
num_people_3 = int(people)
total_amount_3 = meal_cost + service_fee - discount_3
per_person_cost_3 = total_amount_3 / num_people_3
is_over_budget_3 = per_person_cost_3 > budget_per_person
print(f"折價券：{discount_3} 元")
print(f"應付總額：{total_amount_3:.2f} 元")
print(f"每人費用：{per_person_cost_3:.2f} 元")
print(f"是否超過預算：{is_over_budget_3}")
```
