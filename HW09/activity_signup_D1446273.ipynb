```python
# HW09：Python 資料型別上機實作：兩場活動報名整理器
# 姓名：廖音淇
# 學號：D1446273

# 固定的活動名稱 (tuple) 與每人費用 (dict)
activity_names = ("桌遊體驗", "電影欣賞")
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}
board_name, movie_name = activity_names

# ==========================================
# 測試一：原始資料
# ==========================================
print("【測試一：原始資料】")
board_raw = " Amy, Ben, amy, Cora "
movie_raw = " ben, Dora, ERIC, ben "

# 1. 整理字串並轉成保留重複的 list
board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

# 2. 轉成 set 去重與集合運算
board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set          # 交集：兩場都有參加
all_set = board_set | movie_set           # 聯集：至少參加一場
only_board_set = board_set - movie_set    # 差集：只參加桌遊
only_movie_set = movie_set - board_set    # 差集：只參加電影

# 3. 排序與切片取得前兩位
all_sorted = sorted(all_set)
notify_preview = all_sorted[:2]

# 4. 使用 dict 保存摘要與計算金額
summary_1 = {
    "桌遊人數": len(board_set),
    "電影人數": len(movie_set),
    "共同人數": len(both_set),
    "總不重複人數": len(all_set),
    "報名人次": len(board_set) + len(movie_set),
    "桌遊應收": len(board_set) * unit_fees[board_name],
    "電影應收": len(movie_set) * unit_fees[movie_name],
}
summary_1["總應收"] = summary_1["桌遊應收"] + summary_1["電影應收"]

# 5. 輸出摘要
print(f"活動：{board_name}，每人 {unit_fees[board_name]:.2f} 元")
print(f"活動：{movie_name}，每人 {unit_fees[movie_name]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{sorted(board_set)}")
print(f"電影去重名單：{sorted(movie_set)}")
print(f"桌遊人數：{summary_1['桌遊人數']} 人")
print(f"電影人數：{summary_1['電影人數']} 人")
print(f"共同參加者：{sorted(both_set)}")
print(f"共同人數：{summary_1['共同人數']} 人")
print(f"只參加桌遊者：{sorted(only_board_set)}")
print(f"只參加電影者：{sorted(only_movie_set)}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary_1['總不重複人數']} 人")
print(f"報名人次：{summary_1['報名人次']} 人次")
print(f"通知名單前兩位：{notify_preview}")
print(f"桌遊應收：{summary_1['桌遊應收']:.2f} 元")
print(f"電影應收：{summary_1['電影應收']:.2f} 元")
print(f"總應收：{summary_1['總應收']:.2f} 元\n")


# ==========================================
# 測試二：新增一位電影參加者 (Frank)
# ==========================================
print("【測試二：新增一位電影參加者】")
board_raw_2 = " Amy, Ben, amy, Cora "
movie_raw_2 = " ben, Dora, ERIC, ben, Frank "

board_list_2 = board_raw_2.replace(" ", "").lower().split(",")
movie_list_2 = movie_raw_2.replace(" ", "").lower().split(",")

board_set_2 = set(board_list_2)
movie_set_2 = set(movie_list_2)

both_set_2 = board_set_2 & movie_set_2
all_set_2 = board_set_2 | movie_set_2
only_board_set_2 = board_set_2 - movie_set_2
only_movie_set_2 = movie_set_2 - board_set_2

all_sorted_2 = sorted(all_set_2)
notify_preview_2 = all_sorted_2[:2]

summary_2 = {
    "桌遊人數": len(board_set_2),
    "電影人數": len(movie_set_2),
    "共同人數": len(both_set_2),
    "總不重複人數": len(all_set_2),
    "報名人次": len(board_set_2) + len(movie_set_2),
    "桌遊應收": len(board_set_2) * unit_fees[board_name],
    "電影應收": len(movie_set_2) * unit_fees[movie_name],
}
summary_2["總應收"] = summary_2["桌遊應收"] + summary_2["電影應收"]

print(f"桌遊人數：{summary_2['桌遊人數']} 人 | 電影人數：{summary_2['電影人數']} 人")
print(f"共同參加者：{sorted(both_set_2)}（共 {summary_2['共同人數']} 人）")
print(f"只參加桌遊者：{sorted(only_board_set_2)}")
print(f"只參加電影者：{sorted(only_movie_set_2)}")
print(f"總不重複人數：{summary_2['總不重複人數']} 人 | 報名人次：{summary_2['報名人次']} 人次")
print(f"通知名單前兩位：{notify_preview_2}")
print(f"桌遊應收：{summary_2['桌遊應收']:.2f} 元 | 電影應收：{summary_2['電影應收']:.2f} 元 | 總應收：{summary_2['總應收']:.2f} 元\n")


# ==========================================
# 測試三：兩場去重後完全相同
# ==========================================
print("【測試三：兩場去重後完全相同】")
board_raw_3 = " Amy, Ben, amy, Cora "
movie_raw_3 = " AMY, ben, CORA, amy "

board_list_3 = board_raw_3.replace(" ", "").lower().split(",")
movie_list_3 = movie_raw_3.replace(" ", "").lower().split(",")

board_set_3 = set(board_list_3)
movie_set_3 = set(movie_list_3)

both_set_3 = board_set_3 & movie_set_3
all_set_3 = board_set_3 | movie_set_3
only_board_set_3 = board_set_3 - movie_set_3
only_movie_set_3 = movie_set_3 - board_set_3

all_sorted_3 = sorted(all_set_3)
notify_preview_3 = all_sorted_3[:2]

summary_3 = {
    "桌遊人數": len(board_set_3),
    "電影人數": len(movie_set_3),
    "共同人數": len(both_set_3),
    "總不重複人數": len(all_set_3),
    "報名人次": len(board_set_3) + len(movie_set_3),
    "桌遊應收": len(board_set_3) * unit_fees[board_name],
    "電影應收": len(movie_set_3) * unit_fees[movie_name],
}
summary_3["總應收"] = summary_3["桌遊應收"] + summary_3["電影應收"]

print(f"桌遊人數：{summary_3['桌遊人數']} 人 | 電影人數：{summary_3['電影人數']} 人")
print(f"共同參加者：{sorted(both_set_3)}（共 {summary_3['共同人數']} 人）")
print(f"只參加桌遊者：{sorted(only_board_set_3)}")
print(f"只參加電影者：{sorted(only_movie_set_3)}")
print(f"總不重複人數：{summary_3['總不重複人數']} 人 | 報名人次：{summary_3['報名人次']} 人次")
print(f"通知名單前兩位：{notify_preview_3}")
print(f"桌遊應收：{summary_3['桌遊應收']:.2f} 元 | 電影應收：{summary_3['電影應收']:.2f} 元 | 總應收：{summary_3['總應收']:.2f} 元")
```
